```python
import pandas as pd
from mEncoder import fig_dir
import matplotlib.pyplot as plt
import matplotlib

MODEL = "EGFR_MAPK"
DATA = "dream_cytof"
SAMPLES = "0_5"

outdir = fig_dir / MODEL / DATA

%matplotlib inline
```

```python
dfs = []

for dataset in ["train", "test"]:
    pretraining = pd.read_csv(
        outdir / f"{SAMPLES}_evaluate_pretraining_cross_sample_{dataset}.csv",
        index_col=[0],
    )
    training = pd.read_csv(
        outdir / f"{SAMPLES}_evaluate_training_{dataset}.csv", index_col=[0]
    )

    dfd = pd.concat([training, pretraining])
    dfd = dfd[dfd["sample"].apply(lambda x: x.endswith("_dyn"))]
    dfd["dataset"] = dataset
    dfs.append(dfd)

df = pd.concat(dfs).reset_index()
```

```python
import seaborn as sns

g = sns.FacetGrid(data=df, col="dataset")
g.map_dataframe(sns.lineplot, x="alpha", y="chi2", hue="layers", style="type")
g.set(yscale="log", xscale="log")
g.add_legend()
```
