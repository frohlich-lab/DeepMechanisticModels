```python
from common import data_dir
import pandas as pd

cytof = pd.read_csv(data_dir / "cytof.csv", index_col=0)
median_phospho = pd.read_csv("median_phospho_data.csv")
```

```python
cytof["treatment"] = cytof["simulationConditionId"].apply(lambda x: x.split("__")[1])
```

```python
cytof_wide = cytof.pivot_table(
    index=["preequilibrationConditionId", "time", "treatment"],
    columns="FEATURE_ID",
    values="measurement",
    aggfunc="mean",
).reset_index()
```

```python
cytof_wide
```

```python
median_phospho
```

```python
cytof_egf_zero = cytof_wide[(cytof_wide.treatment == "EGF") & (cytof_wide.time == 0)]
cytof_egf_zero.drop(columns=["glob_cellID"], inplace=True)
cytof_egf_zero.rename(columns={"preequilibrationConditionId": "cell_line"}, inplace=True)
cytof_egf_zero.set_index("cell_line", inplace=True)
cytof_egf_zero.sort_index(inplace=True)
cytof_egf_zero
```

```python
median_phospho_egf_zero = median_phospho[(median_phospho.treatment == "EGF") & (median_phospho.time == 0)]
median_phospho_egf_zero["cell_line"] = median_phospho_egf_zero["cell_line"].apply(lambda x: "c"+x)
median_phospho_egf_zero.set_index("cell_line", inplace=True)
median_phospho_egf_zero.sort_index(inplace=True)
```

```python
markers = [col for col in cytof_egf_zero.columns if col not in ["preequilibrationConditionId", "time", "treatment"]]
```

```python
diffs = {}
for marker in markers:
    if marker not in median_phospho_egf_zero.columns:
        print(marker)
    else:
        diffs[marker] = cytof_egf_zero[marker].values - median_phospho_egf_zero[marker].values
```

```python
pd.DataFrame(diffs, index=median_phospho_egf_zero.index).abs()
```
